```python
import os

from  dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
gpt4o_llm = ChatOpenAI(model="gpt-4o", temperature=0)
response = gpt4o_llm.invoke("Hola, cómo estás?")
response.pretty_print()
```

```python
from langchain_openai import ChatOpenAI
gpt3_llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
response = gpt3_llm.invoke("Hola, cómo estás?")
response.pretty_print()
```

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

msg_1 = SystemMessage(content="You are a helpful assistant.")
msg_2 = HumanMessage(content="Me llamo Carolina")
msg_3 = AIMessage(content="Hola Carolina, ¿en qué puedo ayudarte?")
msg_4 = HumanMessage(content="Cómo me llamo?")

history = [msg_1, msg_2, msg_3, msg_4]
response = gpt3_llm.invoke(history)
response.pretty_print()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-4o", temperature=0)
response = llm.invoke(history)
response.pretty_print()
```
