```python
from langchain_core.messages import AIMessage

ai_msg = AIMessage(content="Hello, Message from the AI!")
ai_msg.text
```

```python
from langchain_core.messages import HumanMessage

human_msg = HumanMessage(content="Hello, Message from the Human!")
human_msg.text
```

```python
history = [ai_msg, human_msg]
for m in history:
    m.pretty_print()
```
