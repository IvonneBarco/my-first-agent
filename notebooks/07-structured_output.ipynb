```python
import os
from  dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-4o", temperature=0)
response = llm.invoke("Hola, soy Ivonne y mi telefono es 1234567890.")
response.text
```

```python
from pydantic import BaseModel, Field
class ContactInfo(BaseModel):
    name: str = Field(description="The full name of the person")
    email: str = Field(description="The email address of the person")
    phone: str = Field(description="The phone number of the person")
    tone: int = Field(description="The tone of the person", ge=0, le=100)
    age: int = Field(description="The age of the person")
    sentiment: str = Field(description="The sentiment of the conversation")

llm_with_structured_output = llm.with_structured_output(schema=ContactInfo)

messages = [
    ("system", "You are a helpful assistant that extracts contact information from a given conversation."),
    ("user", "Hola, soy Ivonne y mi telefono es 1234567890.")
]

response = llm_with_structured_output.invoke(messages)
response
```
